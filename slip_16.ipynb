```python
import numpy as np
import pandas as pd
import matplotlib.pyplot
```

```python
dataset=pd.read_csv("slip_16.csv",header=None)
tran=[]
for i in range(0,7501):
    tran.append([str(dataset.values[i,j]) for j in range(0,20)if str(dataset.values[i,j]) != 'nan'])
#conver to string cux they are product namesEnsures all items are strings (no float NaNs)
```

```python
from apyori import apriori
rules=apriori(transactions=tran,min_support=0.003,min_confidence=0.2,min_lift=3,min_length=2,max_length=2)

#we train the model over here
```

```python
res=list(rules)
print(res)
#we convert it to a list cuz this returns generator object
```

```python
def inspect(res):
    lhs    =[tuple(ress[2][0][0])[0] for ress in res]
    rhs    =[tuple(ress[2][0][1])[0] for ress in res]
    support  =[ress[1] for ress in res]
    con      =[ress[2][0][2] for ress in res]
    lift=[ress[2][0][3] for ress in res]
    return (zip(lhs,rhs,support,con,lift))
resultsinDataFrame=pd.DataFrame(inspect(res),columns=["Left hand side","Right hand side","Suppprt","Confidence","Lift"])
```

```python
print(resultsinDataFrame.to_string())
```

```python
resultsinDataFrame.nlargest(n=10, columns="Lift")
# Among these, Lift is often considered one of the most important metrics for finding strong and interesting rules, because:

# Lift > 1 → Positive association between antecedent and consequent.

# Lift ≈ 1 → No association (they appear independently).

# Lift < 1 → Negative association (they appear less often together than expected).
```
