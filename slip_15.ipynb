```python
# Step 1: Import required libraries
import numpy as np
import pandas as pd
from apyori import apriori

# Step 2: Load the dataset
dataset = pd.read_csv("slip_15.csv")

# Step 3: Generate list of transactions (similar style to your original code)
tran = []
members = dataset["Member_number"].unique()  # get unique member IDs

for m in members:
    items = dataset[dataset["Member_number"] == m]["itemDescription"].values
    # convert each item to string and ignore NaNs
    tran.append([str(items[i]) for i in range(len(items)) if str(items[i]) != 'nan'])

# Step 4: Train Apriori model
rules = apriori(
    transactions=tran,
    min_support=0.002,     # minimum support threshold
    min_confidence=0.2,    # minimum confidence threshold
    min_lift=3,            # minimum lift threshold
    min_length=2,          # minimum number of items in a rule
    max_length=2           # maximum number of items in a rule
)

# Step 5: Convert generator to list
res = list(rules)
print(res)  # prints all raw rules

# Step 6: Extract rules into a DataFrame
def inspect(res):
    lhs = [tuple(ress[2][0][0])[0] for ress in res]
    rhs = [tuple(ress[2][0][1])[0] for ress in res]
    support = [ress[1] for ress in res]
    con = [ress[2][0][2] for ress in res]
    lift = [ress[2][0][3] for ress in res]
    return zip(lhs, rhs, support, con, lift)

resultsinDataFrame = pd.DataFrame(
    inspect(res),
    columns=["Left hand side", "Right hand side", "Support", "Confidence", "Lift"]
)

# Step 7: Print all rules
print(resultsinDataFrame.to_string())

# Step 8: Top 10 rules by Lift
print("\nTop 10 rules by Lift:")
print(resultsinDataFrame.nlargest(10, "Lift"))
```
