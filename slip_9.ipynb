```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dataset=pd.read_csv('slip_9.csv')
print(dataset.head())
print(dataset.isnull().sum())
```

```python
X=dataset.iloc[:,[1]].values
y=dataset.iloc[:,[5]].values
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=0)
```

```python
from sklearn.linear_model import LinearRegression
le=LinearRegression()
le.fit(X_train,y_train)
```

```python
y_pred=le.predict(X_test)
```

```python
plt.scatter(X_train,y_train,color='red')
plt.plot(X_train,le.predict(X_train),color='blue')
plt.title('Glucose vs BMI (Training set)')
plt.xlabel('Glucose')
plt.ylabel('BMI')
plt.show()
```

```python
plt.scatter(X_test,y_test,color='red')
plt.plot(X_train,le.predict(X_train),color='blue')
plt.title('Glucose vs BMI (Testing set)')
plt.xlabel('Glucose')
plt.ylabel('BMi')
plt.show()
```

```python
print('coefficient : ',le.coef_)
print('intercept : ',le.intercept_)
```

```python
res=np.sum((y_test-y_pred)**2)
print('Residual sum of square : ',res)
```

```python
from sklearn.metrics import r2_score
r2=r2_score(y_test,y_pred)
print('Coefficient of determination : ',r2)
```
