```python
import pandas as pd
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler

print("--- Problem 24: SVM for User Behavior (user_behavior_dataset.csv) ---")

# Load dataset
data = pd.read_csv('slip_24.csv')

# --- 1. Pre-processing: Convert text to numbers ---
# Drop the ID, as it's not a feature
data = data.drop('User ID', axis=1)

# Use get_dummies to one-hot encode text columns
data_processed = pd.get_dummies(data, columns=['Device Model', 'Operating System', 'Gender'], drop_first=True)

# --- 2. Define Features (X) and Target (y) ---
X = data_processed.drop('User Behavior Class', axis=1)
y = data_processed['User Behavior Class']

# --- 3. Pre-processing: Scale the features ---
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# --- 4. Split data into training and testing sets ---
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# --- 5. Implement SVM Classifier ---
model = SVC(random_state=42)
model.fit(X_train, y_train)

# --- 6. Evaluate the performance ---
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)

print(f"Model Accuracy: {accuracy * 100:.2f}%")
```
