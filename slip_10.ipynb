```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dataset=pd.read_csv('slip_10.csv')
```

```python
X=dataset.iloc[:,[0]].values
y=dataset.iloc[:,[1]].values
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
X_train=sc.fit_transform(X_train)
X_test=sc.transform(X_test)
```

```python
from sklearn.linear_model import LinearRegression
le=LinearRegression()
le.fit(X_train,y_train)
```

```python
y_pred=le.predict(X_test)
```

```python
plt.scatter(X_train, y_train, color='red')
plt.plot(X_train, le.predict(X_train), color='blue')
plt.title('SAT vs GPA (Training set)')
plt.xlabel('SAT Score')
plt.ylabel('GPA')
plt.show()
```

```python
plt.scatter(X_test, y_test, color='red')
plt.plot(X_train, le.predict(X_train), color='blue')
plt.title('SAT vs GPA (Testing set)')
plt.xlabel('SAT Score')
plt.ylabel('GPA')
plt.show()
```

```python
print('Coefficient : ',le.coef_)
print('Intercept : ',le.intercept_)
```

```python
res=np.sum((y_test-y_pred)**2)
print('Residual sum of squares : ',res)
```

```python
from sklearn.metrics import r2_score
r2=r2_score(y_test,y_pred)
print('Coefficient of Determination : ',r2)
```
