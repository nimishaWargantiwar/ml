```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dataset=pd.read_csv('slip_6.csv')
print(dataset.columns)
X=dataset.iloc[:,[10,6]].values
```

```python
import scipy.cluster.hierarchy as sch
sch=sch.dendrogram(sch.linkage(X,method='ward'))
plt.title('Dendrogram')
plt.xlabel('Players')
plt.ylabel('Euclidean dist')
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering
c=AgglomerativeClustering(n_clusters=2,linkage='ward',metric='euclidean')
y_pred=c.fit_predict(X)
```

```python
plt.scatter(X[y_pred==0,0],X[y_pred==0,1],s=100,c='Red',label='Cluster 1')
plt.scatter(X[y_pred==1,0],X[y_pred==1,1],s=100,c='Green',label='Cluster 2')
plt.title('Clusters')
plt.xlabel('Cruns')
plt.ylabel('Years')
plt.legend()
plt.show()
```

```python
from sklearn.metrics import silhouette_score
s=silhouette_score(X,y_pred)
print(s)
```
