```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dataset=pd.read_csv('slip_8.csv')
print(dataset.columns)
dataset.head(10)
```

```python

print(dataset.isnull().sum())
```

```python
X=dataset.iloc[:,[4,0]].values
```

```python
import scipy.cluster.hierarchy as sch
sch=sch.dendrogram(sch.linkage(X,method='ward'))
plt.title('Dendrogram')
plt.xlabel('Startups')
plt.ylabel('Euclidean dist')
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering
a=AgglomerativeClustering(n_clusters=3,linkage='ward',metric='euclidean')
y_pred=a.fit_predict(X)
```

```python
plt.scatter(X[y_pred==0,0],X[y_pred==0,1],s=100,c='Red',label='Cluster 1')
plt.scatter(X[y_pred==1,0],X[y_pred==1,1],s=100,c='Green',label='Cluster 2')
plt.scatter(X[y_pred==2,0],X[y_pred==2,1],s=100,c='Pink',label='Cluster 3')
plt.title('Clusters')
plt.xlabel("Profit")
plt.ylabel('R&D Spend')
plt.legend()
plt.show()
           
```

```python
from sklearn.metrics import silhouette_score
s=silhouette_score(X,y_pred)
print(s)
```
