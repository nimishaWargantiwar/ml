```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dataset=pd.read_csv('slip_22.csv')
X=dataset.iloc[:,:-1].values
y=dataset.iloc[:,-1].values
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
X_train=sc.fit_transform(X_train)
X_test=sc.transform(X_test)
```

```python
from sklearn.svm import SVC

from sklearn.metrics import accuracy_score, confusion_matrix
kernels = ['linear', 'poly', 'rbf', 'sigmoid']
best_acc = 0
best_kernel = ''
accuracies = []   # added this to store accuracies

for kernel in kernels:
    print("\n================== Kernel :", kernel.upper(), "========================")    
    # Create and train the SVM model
    classifier = SVC(kernel=kernel, random_state=0)
    classifier.fit(X_train, y_train)
    
    # Predict
    y_pred = classifier.predict(X_test)
    
    # Confusion matrix and accuracy
    c = confusion_matrix(y_test, y_pred)
    print("Confusion Matrix:")
    print(c)
    
    acc=accuracy_score(y_test, y_pred) * 100
    print("Accuracy:", acc)
    accuracies.append(acc)
    
    # store each accuracy
    
    if acc > best_acc:
            best_acc = acc
            best_kernel = kernel
print("\n Best Performing Kernel:", best_kernel.upper())
print(" Highest Accuracy:", best_acc)
        
```

```python
plt.figure(figsize=(7,7))
bars = plt.bar(kernels, accuracies, color=['skyblue', 'lightgreen', 'orange', 'violet'])
plt.bar_label(bars, padding=3)
plt.title("SVM Kernel Accuracy Comparison (Mobile Price Range Classification)")
plt.xlabel("Kernel Type")
plt.ylabel("Accuracy (%)")
plt.ylim(0, 100)  
plt.show()
```
