```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dataset=pd.read_csv('slip_7.csv')
X=dataset.iloc[:,[3,2]].values
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
X=sc.fit_transform(X)
```

```python
from sklearn.cluster import KMeans
wcss=[]
for i in range(1,11):
    kmeans=KMeans(n_clusters=i,random_state=42,init='k-means++')
    kmeans.fit(X)
    wcss.append(kmeans.inertia_)
plt.plot(range(1,11),wcss)
plt.title('Elbow method')
plt.xlabel('clusters')
plt.ylabel('WCSS')
plt.show()
```

```python
kmeans=KMeans(n_clusters=6,random_state=42,init='k-means++')
kmeans.fit(X)
```

```python
y_pred=kmeans.predict(X)
```

```python
plt.scatter(X[y_pred==0,0],X[y_pred==0,1],s=100,c='Red',label='Cluster 1')
plt.scatter(X[y_pred==1,0],X[y_pred==1,1],s=100,c='Green',label='Cluster 1')
plt.scatter(X[y_pred==2,0],X[y_pred==2,1],s=100,c='Blue',label='Cluster 2')
plt.scatter(X[y_pred==3,0],X[y_pred==3,1],s=100,c='Orange',label='Cluster 3')
plt.scatter(X[y_pred==4,0],X[y_pred==4,1],s=100,c='Pink',label='Cluster 4')
plt.scatter(X[y_pred==5,0],X[y_pred==5,1],s=100,c='Cyan',label='Cluster 5')
plt.scatter(kmeans.cluster_centers_[:,0],kmeans.cluster_centers_[:,1],c='black',s=200,label='Centroids')
plt.title('Clusters')
plt.xlabel('Estimated Salary')
plt.ylabel('Age')
plt.legend()
plt.show()
```

```python
from sklearn.metrics import silhouette_score
s=silhouette_score(X,y_pred)
print(round(s,2))
```
