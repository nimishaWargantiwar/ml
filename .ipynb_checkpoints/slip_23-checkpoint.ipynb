```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dataset=pd.read_csv('slip_23.csv')
X=dataset.iloc[:,dataset.columns!='Personal Loan'].values
y=dataset.iloc[:,dataset.columns=='Personal Loan'].values.ravel()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
X_train=sc.fit_transform(X_train)
X_test=sc.transform(X_test)
```

```python
from sklearn.svm import SVC
from sklearn.metrics import confusion_matrix,accuracy_score

kernels=['linear','rbf','poly','sigmoid']
best_acc=0
best_acc_kernel=''
accuracies=[]
for k in kernels:
    print("\n =========================== Kernel : ",k.upper()," =============================")
    c=SVC(kernel=k,random_state=0)
    c.fit(X_train,y_train)

    y_pred=c.predict(X_test)

    c=confusion_matrix(y_test,y_pred)
    print("Confusion matrix : \n",c)

    acc=accuracy_score(y_test,y_pred)*100
    #print(f"Accuracy : {acc:.2f}%")
    print("Accuracy : ",acc,"%")
    accuracies.append(acc)

    if acc>best_acc:
        best_acc=acc
        best_acc_kernel=k

print("\nBest kernel : ",best_acc_kernel.upper())
print("Accuracy : ",best_acc,"%")
    
```

```python
plt.figure(figsize=(9,12))
bars=plt.bar(kernels,accuracies,color=['skyblue','yellow','violet','pink'])
plt.bar_label(bars,padding=3)
plt.title('SVM Kernel Accuracy')
plt.xlabel('Kernel Type')
plt.ylabel('Accuracy')
plt.ylim(0,100)
plt.show()
```
