```python
import numpy as np

# Points
points = np.array([
    [0.1, 0.6],
    [0.15, 0.71],
    [0.08, 0.9],
    [0.16, 0.85],
    [0.2, 0.3],
    [0.25, 0.5],
    [0.24, 0.1],
    [0.3, 0.2]
])

# Initial centroids
m1 = points[0]  # P1
m2 = points[7]  # P8

# Assign points to nearest centroid
clusters = {1: [], 2: []}
for idx, p in enumerate(points):
    d1 = np.linalg.norm(p - m1)
    d2 = np.linalg.norm(p - m2)
    if d1 < d2:
        clusters[1].append(idx+1)
    else:
        clusters[2].append(idx+1)

# Update centroids
m1_new = np.mean(points[[i-1 for i in clusters[1]]], axis=0)
m2_new = np.mean(points[[i-1 for i in clusters[2]]], axis=0)

print("Clusters assignment:", clusters)
print("Updated m1:", m1_new)
print("Updated m2:", m2_new)
print("Population around m2:", len(clusters[2]))
```
